## Fat per dollar

This analysis looks at fat per dollar for different types of cheese sold at Walmart and Wholefoods, using [GroceryDB](https://github.com/Barabasi-Lab/GroceryDB) data. 

### 1. Importing and setting up


```python
import pandas as pd

df = pd.read_csv("GroceryDB_foods.csv")
```

### 2. Creating a subset of cheese data with name of the cheese, price, total fat and store, columns.

```python
df_cheese = df.loc[
    df["harmonized single category"] == "cheese",
    ["harmonized single category", "name", "price", "Total Fat","store"]
]
```

### 3. Defining cheese categories

```python
def categorize_cheese(name):
    name_lower = str(name).lower()

    if 'cheddar' in name_lower:
        return 'Cheddar'
    elif 'mozzarella' in name_lower:
        return 'Mozzarella'
    elif 'parmesan' in name_lower or 'parmigiano' in name_lower:
        return 'Parmesan'
    elif 'cream cheese' in name_lower or 'philadelphia' in name_lower:
        return 'Cream Cheese'
    elif 'goat' in name_lower or 'chevre' in name_lower:
        return 'Goat'
    elif 'brie' in name_lower:
        return 'Brie'
    elif 'feta' in name_lower:
        return 'Feta'
    elif 'gouda' in name_lower:
        return 'Gouda'
    elif 'blue' in name_lower or 'bleu' in name_lower:
        return 'Blue'
    elif 'american' in name_lower:
        return 'American'
    elif 'cottage' in name_lower:
        return 'Cottage'
    elif 'swiss' in name_lower:
        return 'Swiss'
    elif 'colby' in name_lower:
        return 'Colby'
    elif 'provolone' in name_lower:
        return 'Provolone'
    elif 'pepper jack' in name_lower:
        return 'Pepper Jack'
    elif 'ricotta' in name_lower:
        return 'Ricotta'
    else:
        return None
```

### 4. Categorising cheese by type

```python

df_cheese["cheese_type"] = df_cheese["name"].apply(categorize_cheese)
df_cheese
```

### 5. Removing the NAs

```python
df_cheese = df_cheese[df_cheese['cheese_type'].notna()].copy()
df_cheese
```

### 6. Selecting cheese sold only at Walmart and Whole Foods


```python
df_cheese = df_cheese[df_cheese['store'].isin(['Walmart', 'WholeFoods'])]
df_cheese
```

### 7. filtering the cheese with fat content between 0 and 50

```python
df_cheese = df_cheese[(df_cheese['Total Fat'] >= 0) & (df_cheese['Total Fat'] < 50)]
df_cheese
```

### 8. Copying price and total fat into a new dataset and then filtering price

```python
priceVfat = df_cheese.dropna(subset=['price', 'Total Fat']).copy()
priceVfat = priceVfat [priceVfat ['price'] > 0]
priceVfat
```

### 9.Calculating fat per dollar = Total Fat/price

```python
priceVfat['fat_per_dollar'] = priceVfat['Total Fat'] / priceVfat['price']
priceVfat
```

### 10. Taking the median fat per dollar value for each cheese type

```python
priceVfat = priceVfat.groupby('cheese_type').agg(
      fat_per_dollar=('fat_per_dollar', 'median')).sort_values('fat_per_dollar', ascending= False)
priceVfat
```

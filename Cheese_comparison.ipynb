## Cheese price divide

This analysis looks at the price difference for different types of cheese sold at Walmart and Wholefoods, using [GroceryDB](https://github.com/Barabasi-Lab/GroceryDB) data. 

### 1. Importing and setting up


```python
import pandas as pd
import numpy as np

df = pd.read_csv("GroceryDB_foods.csv")
```

### 2. Creating a subset of cheese data with name of the cheese, price, total fat and store, columns.

```python

df_cheese = df.loc[
    df["harmonized single category"] == "cheese",
    ["harmonized single category", "name", "store", "price"]
]
df_cheese
```

### 3. Defining cheese categories

```python
def categorize_cheese(name):
    name_lower = str(name).lower()

    if 'cheddar' in name_lower:
        return 'Cheddar'
    elif 'mozzarella' in name_lower:
        return 'Mozzarella'
    elif 'parmesan' in name_lower or 'parmigiano' in name_lower:
        return 'Parmesan'
    elif 'cream cheese' in name_lower or 'philadelphia' in name_lower:
        return 'Cream Cheese'
    elif 'goat' in name_lower or 'chevre' in name_lower:
        return 'Goat'
    elif 'brie' in name_lower:
        return 'Brie'
    elif 'feta' in name_lower:
        return 'Feta'
    elif 'gouda' in name_lower:
        return 'Gouda'
    elif 'blue' in name_lower or 'bleu' in name_lower:
        return 'Blue'
    elif 'american' in name_lower:
        return 'American'
    elif 'cottage' in name_lower:
        return 'Cottage'
    elif 'swiss' in name_lower:
        return 'Swiss'
    elif 'colby' in name_lower:
        return 'Colby'
    elif 'provolone' in name_lower:
        return 'Provolone'
    elif 'pepper jack' in name_lower:
        return 'Pepper Jack'
    elif 'ricotta' in name_lower:
        return 'Ricotta'
    else:
        return None
```

### 4. Categorising cheese by type

```python
df_cheese["cheese_type"] = df_cheese["name"].apply(categorize_cheese)
```

### 5. Removing the NAs

```python
df_cheese = df_cheese[df_cheese['cheese_type'].notna()]
df_cheese
```

### 6. Filter to include cheese from only Walmart and Whole Foods


```python
df_cheese = df_cheese[df_cheese['store'].isin(['Walmart', 'WholeFoods'])]
df_cheese
```

### 7. Group into cheese types, stores and median price

```python
df_cheese = df_cheese.groupby(['cheese_type','store'])['price'].median().unstack()
```

### 8. Remove NA's again

```python
df_cheese= df_cheese.dropna()
```

### 9. Calculating the price difference and percentage price difference for Walmart Vs Wholefoods

```python
df_cheese['difference'] = (df_cheese['WholeFoods'] - df_cheese['Walmart']).round(2)
df_cheese['pct_difference'] = (((df_cheese['WholeFoods'] / df_cheese['Walmart']) - 1) * 100).round(2)
df_cheese = df_cheese.sort_values('difference', ascending=False)
df_cheese
```

### 10. Calculating the median Price difference

```python
median = df_cheese['difference'].median()
median
```
